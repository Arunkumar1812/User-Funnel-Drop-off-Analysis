```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
largest= df.loc[df['drop_off'].idxmax()]
print(largest)
```

```python
plt.Figure(figsize=(100,100))

plt.bar(df['step'], df['users'])

plt.title("user funnel")
plt.xlabel("stage" )
plt.ylabel("users")
plt.xticks(rotation=20)
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.plot(
    df['step'],
    df['conversion_rate'],
    marker='o',
    markersize=10,
    color='red'
)

plt.title("Conversion Rate")
plt.xlabel("Stages")
plt.ylabel("Conversion Rate (%)")

plt.grid(axis='y', color="brown", linestyle="dotted" )

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(df['step'], df['drop_off'])

plt.title("Drop-off Rate")
plt.xlabel("Stages")
plt.ylabel("Drop-off (%)")

plt.xticks(rotation=20)
plt.show()
```

```python
plt.figure(figsize=(7,7))

plt.pie(df['users'],
        labels=df['step'],
        autopct='%1.1f%%',
        startangle=90)

plt.title("User Distribution Across Funnel Stages")

plt.show()
```

```python
print(" Funnel Analysis ")

largest = df.loc[df['drop_off'].idxmax()]

print(f"Biggest Drop-off Stage : {largest['step']}")
print(f"Users at this stage    : {largest['users']}")
print(f"Conversion Rate        : {largest['conversion_rate']}%")
print(f"Drop-off Rate          : {largest['drop_off']}%")
```
